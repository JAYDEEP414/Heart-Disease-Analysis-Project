# HEART DISEASE DATA CLEANING + EDA PROJECT

# 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Display all columns

```python
pd.set_option('display.max_columns', None)
```

# 2. Load Dataset

```python
df = pd.read_csv("Heart_Disease_Dataset.csv")
```

# 3. Display First 5 Rows

```python
print("First 5 Rows of Dataset:")
display(df.head())
```

# 4. Dataset Shape

```python
print("Dataset Shape (Rows, Columns):", df.shape)
```

# 5. Column Names

```python
print("\nColumn Names:")
print(df.columns.tolist())
```

# 6. Dataset Information

```python
print("\nDataset Info:")
df.info()
```

# 7. Check Missing Values

```python
print("\nMissing Values in Each Column:")
print(df.isnull().sum())
```

# 8. Check Duplicate Rows

```python
print("\nNumber of Duplicate Rows:", df.duplicated().sum())
```

# 9. Summary Statistics

```python
print("\nSummary Statistics:")
display(df.describe(include='all'))
```

# 10. Separate Numeric and Categorical Columns

```python
numeric_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(exclude=np.number).columns

print("\nNumeric Columns:")
print(list(numeric_cols))

print("\nCategorical Columns:")
print(list(categorical_cols))
```

# 11. Remove Duplicate Rows

```python
df.drop_duplicates(inplace=True)
print("\nShape After Removing Duplicates:", df.shape)
```

# 12. Handle Missing Values

```python
for col in numeric_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

# 13. Standardize Text Columns

```python
for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip().str.title()
```

# 14. Convert Target Column to Integer

```python
if 'Heart_Disease' in df.columns:
    df['Heart_Disease'] = pd.to_numeric(df['Heart_Disease'], errors='coerce')
    df['Heart_Disease'] = df['Heart_Disease'].fillna(df['Heart_Disease'].mode()[0])
    df['Heart_Disease'] = df['Heart_Disease'].astype(int)
```

# 15. Outlier Detection Using Boxplots

```python
print("\nGenerating Boxplots for Numeric Columns...")
for col in numeric_cols[:6]:  # Show first 6 numeric columns
    plt.figure(figsize=(6, 3))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

# 16. Heart Disease Distribution

```python
if 'Heart_Disease' in df.columns:
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x='Heart_Disease')
    plt.title("Heart Disease Distribution")
    plt.xlabel("Heart Disease (0 = No Disease, 1 = Disease)")
    plt.ylabel("Count")
    plt.show()
```

# 17. Age Distribution

```python
if 'Age' in df.columns:
    plt.figure(figsize=(8, 4))
    plt.hist(df['Age'], bins=20)
    plt.title("Age Distribution")
    plt.xlabel("Age")
    plt.ylabel("Frequency")
    plt.show()
```

# 18. Age vs Cholesterol Scatter Plot

```python
if 'Age' in df.columns and 'Cholesterol' in df.columns:
    plt.figure(figsize=(8, 5))
    sns.scatterplot(data=df, x='Age', y='Cholesterol', hue='Heart_Disease')
    plt.title("Age vs Cholesterol")
    plt.show()
```

# 19. Smoking vs Heart Disease

```python
if 'Smoking' in df.columns and 'Heart_Disease' in df.columns:
    print("\nSmoking vs Heart Disease:")
    smoking_analysis = pd.crosstab(df['Smoking'], df['Heart_Disease'])
    display(smoking_analysis)

    smoking_analysis.plot(kind='bar', figsize=(8, 4))
    plt.title("Smoking vs Heart Disease")
    plt.xlabel("Smoking")
    plt.ylabel("Count")
    plt.show()
```

# 20. Correlation Matrix

```python
print("\nGenerating Correlation Matrix...")
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Matrix")
plt.show()
```

# 21. Key Insights

```python
if 'Heart_Disease' in df.columns:
    print("\nHeart Disease Percentage:")
    print(df['Heart_Disease'].value_counts(normalize=True) * 100)

if 'Smoking' in df.columns and 'Heart_Disease' in df.columns:
    print("\nHeart Disease Rate by Smoking:")
    print(df.groupby('Smoking')['Heart_Disease'].mean() * 100)

if 'Diabetes' in df.columns and 'Heart_Disease' in df.columns:
    print("\nHeart Disease Rate by Diabetes:")
    print(df.groupby('Diabetes')['Heart_Disease'].mean() * 100)
```

# 22. Save Cleaned Dataset

```python
df.to_csv("Cleaned_Dataset.csv", index=False)
```
